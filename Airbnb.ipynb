```python
df['address'].values
```

```python
df['address'] = df['address'].astype(str)
```

```python
import ast

df['street'] = df['address'].apply(lambda x: ast.literal_eval(x)['street'])
```

```python
df['suburb'] = df['address'].apply(lambda x: ast.literal_eval(x)['suburb'])
df['government_area'] = df['address'].apply(lambda x: ast.literal_eval(x)['government_area'])
df['market'] = df['address'].apply(lambda x: ast.literal_eval(x)['market'])
df['country'] = df['address'].apply(lambda x: ast.literal_eval(x)['country'])
df['country_code'] = df['address'].apply(lambda x: ast.literal_eval(x)['country_code'])
df['location'] = df['address'].apply(lambda x: ast.literal_eval(x)['location'])

df['location_type'] = df['address'].apply(lambda x: ast.literal_eval(x)['location']['type'])
df['location_exact'] = df['address'].apply(lambda x: ast.literal_eval(x)['location']['is_location_exact'])
df['location_coordinate'] = df['address'].apply(lambda x: ast.literal_eval(x)['location']['coordinates'])
```

```python
# Check for null values in each address column
address_columns = ['street', 'suburb', 'government_area', 'market', 'country', 'country_code', 'location_type', 'location_exact', 'location_coordinate']

for column in address_columns:
    null_count = df[column].isnull().sum()
    print(f"{column}: {null_count} null values")
```
